```python
def add_numbers(*args):
  """Adds an arbitary number of arguments."""
  return sum(args)

  #example usage:
  result = add_numbers(1,2,3,4,5)
  print(f"Sum of numbers (1,2,3,4,5): {result}")

  result_two = add_numbers(10,20,30)
  print(f"sum of numbers (10,20,30): {result_two}")

  result_three = add_numbers(7)
  print(f"sum of number (7): {result_three}")
```
