IT-10

```python

import random

# Three-bin counters
accept_count = 0
rework_count = 0
scrap_count = 0

# Production quantity
TOTAL_PARTS = 100

print("===== JOB 09: QUALITY CONTROL =====")

# Generate and classify parts
for i in range(TOTAL_PARTS):

    diameter = round(random.uniform(24.90, 25.10), 2)

    # Three-bin classification
    if 24.95 <= diameter <= 25.05:
        bin_label = "ACCEPT"
        accept_count += 1

    elif 24.90 <= diameter <= 25.10:
        bin_label = "REWORK"
        rework_count += 1

    else:
        bin_label = "SCRAP"
        scrap_count += 1

    print(
        f"Part {i + 1}: "
        f"{diameter:.2f} mm - {bin_label}"
    )

# Final shift report
print("\n===== SHIFT REPORT =====")
print("Total Parts :", TOTAL_PARTS)
print("ACCEPT      :", accept_count)
print("REWORK      :", rework_count)
print("SCRAP       :", scrap_count)

print("========================")

# Verify total count
assert accept_count + rework_count + scrap_count == TOTAL_PARTS

print("Production tally verified.")
```

```python

import random

# Operator selects production mode
mode = input("Select mode (HALT / SKIP): ").strip().upper()

if mode not in ("HALT", "SKIP"):
    print("Invalid mode. Choose HALT or SKIP.")

else:
    accept_count = 0
    rework_count = 0
    scrap_count = 0
    inspected_count = 0

    total_parts = 100

    print(f"\nProduction started in {mode} mode.")

    for i in range(total_parts):

        diameter = round(random.uniform(24.80, 25.20), 2)
        inspected_count += 1

        # Job 09 three-bin classification
        if 24.95 <= diameter <= 25.05:
            bin_label = "ACCEPT"
            accept_count += 1

        elif 24.90 <= diameter <= 25.10:
            bin_label = "REWORK"
            rework_count += 1

        else:
            bin_label = "SCRAP"
            scrap_count += 1

        print(
            f"Part {i + 1}: "
            f"{diameter:.2f} mm - {bin_label}"
        )

        # HALT mode: stop at first SCRAP
        if bin_label == "SCRAP" and mode == "HALT":
            print("\nSCRAP detected! Production halted.")
            break

        # SKIP mode: record scrap and continue
        elif bin_label == "SCRAP" and mode == "SKIP":
            print("SCRAP logged. Continuing production.")
            continue

    # Final shift report
    print("\n========== SHIFT REPORT ==========")
    print("Production Mode :", mode)
    print("Parts Inspected :", inspected_count)
    print("ACCEPT          :", accept_count)
    print("REWORK          :", rework_count)
    print("SCRAP           :", scrap_count)

    if mode == "HALT" and scrap_count > 0:
        print("Status          : STOPPED AT FIRST SCRAP")
    else:
        print("Status          : BATCH COMPLETED")

    print("==================================")

    assert (
        accept_count + rework_count + scrap_count
        == inspected_count
    )
```

```python

import random

# Production settings
TARGET_ACCEPTS = 5
MAX_ATTEMPTS = 500

consecutive_accepts = 0
total_attempts = 0

print("===== CONSECUTIVE ACCEPT PRODUCTION =====")

# Continue until five consecutive ACCEPTs or safety cap
while (
    consecutive_accepts < TARGET_ACCEPTS
    and total_attempts < MAX_ATTEMPTS
):
    total_attempts += 1

    diameter = round(random.uniform(24.80, 25.20), 2)

    # Three-bin classification
    if 24.95 <= diameter <= 25.05:
        bin_label = "ACCEPT"
        consecutive_accepts += 1

    elif 24.90 <= diameter <= 25.10:
        bin_label = "REWORK"
        consecutive_accepts = 0

    else:
        bin_label = "SCRAP"
        consecutive_accepts = 0

    print(
        f"Part {total_attempts}: "
        f"{diameter:.2f} mm - {bin_label} | "
        f"Consecutive ACCEPTs: {consecutive_accepts}"
    )

# Final production report
print("\n========== SHIFT REPORT ==========")
print("Total Parts Machined :", total_attempts)
print("Target ACCEPTs       :", TARGET_ACCEPTS)
print("Consecutive ACCEPTs  :", consecutive_accepts)

if consecutive_accepts == TARGET_ACCEPTS:
    print("Result               : TARGET ACHIEVED")
else:
    print("Result               : SAFETY CAP REACHED")

print("==================================")

# Verify safety limit
assert total_attempts <= MAX_ATTEMPTS
```
